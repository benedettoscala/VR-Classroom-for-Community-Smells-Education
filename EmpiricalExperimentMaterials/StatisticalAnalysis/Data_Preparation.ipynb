## Merge into one dataset to perform the operations

We added a column at the end indicating from which of the three datasets the original answer was from.

Then, we merge the dataset.

```python
file_name_1 = "Presenza_Pre_Normalizzazione.csv"
file_name_2 = "Remoto_Pre_Normalizzazione.csv"
file_name_3 = "VR_Pre_Normalizzazione.csv"

df_presenza = pd.read_csv(file_path+file_name_1)
df_presenza["Dataset"] = "Presence"

df_remote = pd.read_csv(file_path+file_name_2)
df_remote["Dataset"] = "Remote"

df_vr = pd.read_csv(file_path+file_name_3)
df_vr["Dataset"] = "VR"

# Dataset merge
df = pd.concat([df_presenza, df_remote, df_vr])
df
```

```python
variables = {
    "Learning Outcome": list(range(0, 13)),
    "Self Efficacy": list(range(13, 23)),
    "Immersiveness": list(range(23, 33)),
    "Affective Engagement": list(range(33, 38)),
    "Behavioral Engagement": list(range(38, 44)),
    "Cognitive Engagement": list(range(44, 46)),
    "Disengagement": list(range(46, 49)),
    "NASA-TLX": list(range(49, 55)),
    "NPS": [55]
}

variables_2 = {
    "Learning Outcome": list(range(0, 13)),
    "Self Efficacy": list(range(13, 23)),
    "Immersiveness": list(range(23, 33)),
    "Affective Engagement": list(range(33, 38)),
    "Behavioral Engagement": list(range(38, 44)),
    "Cognitive Engagement": list(range(44, 46)),
    "Disengagement": list(range(46, 49)),
    "Disengagement_Reversed": list(range(49, 52)),   # These are the indices after adding Reversed Disengagement
    "NASA-TLX": list(range(52, 58)),
    "NPS": [58]
}
```

## *Disengagement* Values Reversing

Since the *Disengagement* variable is the only one measured as: [ Good - Bad ], we reversed its scale.

```python
def add_columns_at_position(df, pos, column_names, default_value=None):

    for i, col_name in enumerate(column_names):
        df.insert(pos + i, col_name, default_value)
    return df

new_columns = ['REV - Ero distratt* e non ho seguito la lezione.', 'REV - Durante la lezione ho lasciato che la mia mente vagasse.', 'REV - Ho fatto finta di seguire la lezione.']
df = add_columns_at_position(df, 49, new_columns, default_value=None)

df.iloc[:, variables_2["Disengagement_Reversed"]] = df.iloc[:, variables["Disengagement"]]
df.iloc[:, variables_2["Disengagement_Reversed"]]
```

```python
df.iloc[:, variables_2["Disengagement_Reversed"]] = df.iloc[:, variables_2["Disengagement_Reversed"]].replace({
    1: 7,
    2: 6,
    3: 5,
    4: 4,
    5: 3,
    6: 2,
    7: 1})

df.iloc[:, variables_2["Disengagement_Reversed"]]
```

## Creation of the ID Row

The row contains an ID for each question related to their variable

```python
acronyms = {
    "Learning Outcome": "LO",
    "Self Efficacy": "SE",
    "Immersiveness": "IM",
    "Affective Engagement": "AE",
    "Behavioral Engagement": "BE",
    "Cognitive Engagement": "CE",
    "Disengagement": "DE",
    "Disengagement_Reversed": "RDE",
    "NASA-TLX": "NTLX",
    "NPS": "NPS"
}

question_id = []

for key, value in variables_2.items():
    for i in range(len(value)):
        question_id.append(f"{acronyms[key]}{i + 1}")

question_id.append("Type") # This is for the "Dataset" column which does not correspond to any question
print(question_id)
```

Now we include the array as the first row of the dataset.

```python
question_id_df = pd.DataFrame([question_id], columns=df.columns)

df = pd.concat([question_id_df, df], ignore_index=True)

df
```

## Column Drop

We will use this function to drop the columns based on their ID

```python
def drop_column_by_first_row_value(df, value):
    col_to_drop = df.columns[df.iloc[0] == value]

    if not col_to_drop.empty:
        df = df.drop(columns=col_to_drop)

    return df


#df = drop_column_by_first_row_value(df, "CE1")
```
